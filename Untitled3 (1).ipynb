```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import wavfile


# بارگذاری فایل - نام فایل خود را جایگزین کنید
fs, data = wavfile.read('Recording_2.wav')

# اگر سیگنال استریو است، آن را به مونو تبدیل کنید
if len(data.shape) > 1:
    data = data[:, 0]

# نرمال‌سازی دامنه بین -1 و 1
data = data / np.max(np.abs(data))

# ایجاد محور زمان بر حسب ثانیه
duration = len(data) / fs
time = np.linspace(0, duration, len(data))

# رسم نمودار حوزه زمان (بخش 1-ب)
plt.figure(figsize=(12, 4))
plt.plot(time, data, color='blue')
plt.title('Signal in Time Domain')
plt.xlabel('Time (seconds)')
plt.ylabel('Amplitude')
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ۱. محاسبه تعداد نمونه‌ها
N = len(data) 

# ۲. اعمال تبدیل فوریه (DFT)
X_k = np.fft.fft(data)

# ۳. محاسبه قدر مطلق طیف (Magnitude Spectrum) [cite: 16]
abs_X = np.abs(X_k)
```

```python
# ایجاد محور فرکانس طبق فرمول پروژه
k = np.arange(N)
f_axis = (k / N) * fs

# از آنجایی که طیف فرکانسی سیگنال‌های حقیقی متقارن است، 
# معمولاً فقط نیمه اول (فرکانس‌های مثبت) را رسم می‌کنیم:
half_N = N // 2

plt.figure(figsize=(10, 4))
plt.plot(f_axis[:half_N], abs_X[:half_N])
plt.title("Magnitude Spectrum")
plt.xlabel("Frequency (Hz)")
plt.ylabel("|X[k]|")
plt.grid(True)
plt.show()
```

```python
from scipy import signal

# ۱.۵ برابر سریع‌تر (تعداد نمونه‌ها کاهش می‌یابد)
num_samples_fast = int(len(data) / 1.5)
data_fast = signal.resample(data, num_samples_fast)

# ۰.۷۵ برابر کندتر (تعداد نمونه‌ها افزایش می‌یابد)
num_samples_slow = int(len(data) / 0.75)
data_slow = signal.resample(data, num_samples_slow)
```

```python
def get_spectrum(s, fs_internal):
    N_s = len(s)
    X_s = np.abs(np.fft.fft(s))[:N_s//2]
    f_s = np.linspace(0, fs_internal/2, len(X_s))
    return f_s, X_s

f_orig, X_orig = get_spectrum(data, fs)
f_fast, X_fast = get_spectrum(data_fast, fs)
f_slow, X_slow = get_spectrum(data_slow, fs)

plt.figure(figsize=(12, 6))
plt.plot(f_orig, X_orig/np.max(X_orig), label='Original', alpha=0.7)
plt.plot(f_fast, X_fast/np.max(X_fast), label='Fast (1.5x)', alpha=0.7)
plt.plot(f_slow, X_slow/np.max(X_slow), label='Slow (0.75x)', alpha=0.7)
plt.title("Comparison of Spectra")
plt.xlabel("Frequency (Hz)")
plt.xlim(0, 8000)
plt.legend()
plt.show()
```
